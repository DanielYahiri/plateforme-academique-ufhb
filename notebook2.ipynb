```python
import os

def trouver_type_fichier(rep1, extension):
    # On ajoute le point si l'utilisateur ne l'a pas mis
    ext = '.' + extension if not extension.startswith('.') else extension

    try:
        fichiers = os.listdir(rep1)
        trouve = False

        for nom_fichier in fichiers:
            if nom_fichier.endswith(ext):
                print(nom_fichier)
                trouve = True

        if not trouve:
            print(f"Pas de fichier de type {ext} dans le répertoire {rep1}")

    except FileNotFoundError:
        print("Le répertoire spécifié est introuvable.")

# Utilisation du préfixe 'r' pour le chemin Windows
trouver_type_fichier(r"C:\Donnée base de donnée_CM", "txt")
```

```python
def termes_suite(rang, u_0=2): #le rang de la suite et le termes initial
    U = [0] * (rang + 1)
    U[0]=u_0
    print("la liste des indices de la suite plus petit que 1000 \n" )
    if rang==0 :
        print("U_0","=",U[0])
    else:
        print("U_0","=",U[0])
    for i in range (1,rang + 1) :
        U[i]= U[i-1]**2
        if U[i] <= 1000:
            print("U_",i,"=",U[i])


termes_suite(50)
```

```python
#give the possibility to the user to pick up one of the format datetime
#this is a function to fetch a "date" within a sentence
import datetime as dt

def detectionDate(text, date_format):
    # Split the sentence into individual words
    words = text.split()

    for word in words:
        try:
            # Try to convert each word based on the user's format
            found_date = dt.datetime.strptime(word, date_format)
            return str(found_date.date())
        except :
            # If the word doesn't match the format, move to the next word
            continue

    return "Impossible to find a date matching that format."

# Example Usage:
print(detectionDate('i was born 2022/2/21', "%Y-%m-%d"))
```
